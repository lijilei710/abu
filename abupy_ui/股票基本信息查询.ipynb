```python
# 修复 Python 3.10+ collections 和 scipy 兼容性问题
import sys
if sys.version_info >= (3, 10):
    import collections.abc
    import collections
    for attr in ['Iterable', 'Iterator', 'Mapping', 'MutableMapping', 'MutableSet', 'Sequence', 'MutableSequence', 'Set']:
        if not hasattr(collections, attr):
            setattr(collections, attr, getattr(collections.abc, attr))

# 修复 scipy.interp 导入问题
try:
    from scipy import interp
except ImportError:
    from scipy.interpolate import interp1d as interp
    import scipy
    scipy.interp = interp

# 修复 Python 3.13+ imghdr 移除问题
if sys.version_info >= (3, 13):
    import types
    imghdr = types.ModuleType('imghdr')
    
    def what(file, h=None):
        """简单的图像文件类型检测"""
        if h is None:
            if isinstance(file, str):
                with open(file, 'rb') as f:
                    h = f.read(32)
            else:
                location = file.tell()
                h = file.read(32)
                file.seek(location)
        
        # 检查常见图像格式的魔术字节
        if h.startswith(b'\x89PNG'):
            return 'png'
        elif h.startswith(b'\xff\xd8\xff'):
            return 'jpeg'
        elif h.startswith(b'GIF8'):
            return 'gif'
        elif h.startswith(b'BM'):
            return 'bmp'
        elif h.startswith(b'II\x2a\x00') or h.startswith(b'MM\x00\x2a'):
            return 'tiff'
        return None
    
    imghdr.what = what
    sys.modules['imghdr'] = imghdr

import widget_stock_info
widget_stock_info.show_ui()
```
